### **System Data Preparation**

**Import Module**

All the import module will be written below

```python
import pandas as pd
```

```python
product_df = pd.read_csv("data/olist_products_dataset.csv")
product_df
```

```python
product_translation_df = pd.read_csv("data/product_category_name_translation.csv")
product_translation_df
```

```python
translation_dict = dict(zip(
    product_translation_df['product_category_name'],
    product_translation_df['product_category_name_english']
))
product_df['product_category_name'] = product_df['product_category_name'].map(translation_dict).fillna(product_df['product_category_name'])
product_df
```

```python
order_df = pd.read_csv("data/olist_orders_dataset.csv")
order_df
```

```python
order_items_df = pd.read_csv("data/olist_order_items_dataset.csv")
order_items_df
```

```python
order_items_df = order_items_df[
    order_items_df['order_id'].isin(order_df['order_id'])
]
order_items_df = order_items_df.reset_index(drop=True)
order_items_df
```

```python
order_df = order_df[
    order_df['order_id'].isin(order_items_df['order_id'])
]
order_df = order_df.reset_index(drop=True)
order_df
```

```python
sales_data = order_df.merge(order_items_df, on='order_id')
sales_data
```

```python
quantity_counts = sales_data['product_id'].value_counts()
product_df['quantity'] = product_df['product_id'].map(quantity_counts).fillna(0).astype(int)
product_df
```

```python
product_df.to_csv("system_data/Inventory Data.csv", index=False)
product_df.to_excel(
    "system_data/Inventory Data.xlsx",
    index=False,
    engine="openpyxl"
)
product_df.to_excel(
    "system_data/Inventory Data.xls",
    index=False,
    engine="xlwt"
)
```

```python
sales_data = sales_data[sales_data['order_purchase_timestamp'] >= '2017-02-01']
sales_data
```

```python
customer_data = pd.read_csv("data/olist_customers_dataset.csv")
customer_data = customer_data[['customer_id', 'customer_city', 'customer_state']]
customer_data
```

```python
merge_sales_data = pd.merge(
    sales_data,
    customer_data,
    on='customer_id',
    how='left'
)
merge_sales_data
```

```python
merge_sales_data.to_csv("system_data/Transaction Data.csv", index=False)
merge_sales_data.to_excel(
    "system_data/Transaction Data.xlsx",
    index=False,
    engine="openpyxl"
)
```
